```python
from pathlib import Path
import importlib.metadata
import json
import sys

# Sửa theo Input thực tế; dataset phải có đúng cây thư mục này.
ASSET_ROOT = Path('/kaggle/input/name-that-llm-baseline-assets')
TEST_CSV = Path('/kaggle/input/competitions/name-that-llm/test.csv')
OUTPUT = Path('/kaggle/working/submission.parquet')
ARTIFACT_DIR = ASSET_ROOT / 'artifacts' / 'baseline'
```

```python
assert TEST_CSV.is_file(), f'Missing competition input: {TEST_CSV}'
assert (ASSET_ROOT / 'src' / 'baseline.py').is_file(), 'Attach baseline source under src/'
assert (ARTIFACT_DIR / 'metadata.json').is_file(), 'Attach the fitted baseline artifacts'
metadata = json.loads((ARTIFACT_DIR / 'metadata.json').read_text(encoding='utf-8'))
expected = metadata['environment']['libraries']
actual = {name: importlib.metadata.version(name) for name in expected}
print('Expected versions:', expected)
print('Actual versions:', actual)
assert actual == expected, 'Use matching offline dependencies; see README-baseline.md'
sys.path.insert(0, str(ASSET_ROOT))
from src.baseline import main
main(['predict', '--test', str(TEST_CSV), '--artifact-dir', str(ARTIFACT_DIR), '--output', str(OUTPUT)])
main(['validate', '--test', str(TEST_CSV), '--submission', str(OUTPUT)])
```
